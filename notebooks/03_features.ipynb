```python

from pathlib import Path
import pandas as pd
import numpy as np

# Project directory
project_dir = Path(r"D:\Nexzibyte Solution\nexzibyte-predictive-bi")

# Load the Day 3 cleaned dataset
input_path = project_dir / "data" / "clean" / "Day3_Prospects_Clean.csv"

df = pd.read_csv(input_path)

print("Dataset loaded successfully")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("\nColumn names:")
print(df.columns.tolist())

display(df.head())
```

```python

# Day 5 — Initial Feature Dataset

feature_columns = [
    "Company Name",
    "Industry / Sub-industry",
    "Founded Year (Standardized)",
    "Target Customers / B2B or B2C",
    "Employee Count (Numeric)",
    "Company Size Score",
    "B2B Fit Score",
    "Technology Signal Score",
    "Business Need Score",
    "Recent Business Signal Score",
    "Decision-Maker Score",
    "Data Quality Status",
    "Employee Validation"
]

features_df = df[feature_columns].copy()

print("Feature dataset shape:", features_df.shape)
print("\nMissing values by selected feature:")
display(
    features_df.isna().sum().sort_values(ascending=False)
)

print("\nFeature dataset preview:")
display(features_df.head())

print("\nOriginal dataset unchanged in memory:", df.shape == (180, 28))
```

```python

# Day 5 — Feature Catalogue

feature_catalogue = pd.DataFrame([
    ["Company Name", "Company identifier for reporting",
     "Identifier", "Exclude from model inputs"],
    ["Industry / Sub-industry", "Company's industry category",
     "Categorical", "Encode before modeling"],
    ["Founded Year (Standardized)", "Standardized company founding year",
     "Numeric", "Available; check plausibility"],
    ["Target Customers / B2B or B2C", "Describes the target customer market",
     "Categorical/text", "Standardize or encode"],
    ["Employee Count (Numeric)", "Reported numeric employee count",
     "Numeric", "Check employee evidence and validation"],
    ["Company Size Score", "Existing company-size component score",
     "Numeric score", "Existing score; document its derivation"],
    ["B2B Fit Score", "Existing B2B-fit component score",
     "Numeric score", "Existing score; document its derivation"],
    ["Technology Signal Score", "Existing technology component score",
     "Numeric score", "Existing score; document its derivation"],
    ["Business Need Score", "Existing business-need component score",
     "Numeric score", "Existing score; document its derivation"],
    ["Recent Business Signal Score", "Existing recent-signal component score",
     "Numeric score", "No historical trend data available"],
    ["Decision-Maker Score", "Existing decision-maker component score",
     "Numeric score", "Existing score; document its derivation"],
    ["Data Quality Status", "Indicates record quality-review status",
     "Quality flag", "Use for filtering or audit; avoid as a predictive signal without justification"],
    ["Employee Validation", "Indicates employee-data validation status",
     "Quality flag", "Use for review and audit"]
], columns=[
    "Source Feature",
    "Business Definition",
    "Feature Type",
    "Modeling Notes"
])

display(feature_catalogue)
```

```python

# Day 5 — Save Feature Engineering Outputs

processed_dir = project_dir / "data" / "processed"
processed_dir.mkdir(parents=True, exist_ok=True)

feature_dataset_path = (
    processed_dir / "Day5_Feature_Dataset.csv"
)

feature_catalogue_path = (
    processed_dir / "Day5_Feature_Catalogue.csv"
)

features_df.to_csv(feature_dataset_path, index=False)
feature_catalogue.to_csv(feature_catalogue_path, index=False)

print("Feature dataset saved:", feature_dataset_path)
print("Feature catalogue saved:", feature_catalogue_path)

print("\nSaved dataset shape:", pd.read_csv(feature_dataset_path).shape)
print("Saved catalogue shape:", pd.read_csv(feature_catalogue_path).shape)
```

```python
# Day 5 - Step 4: Engineer company-age feature

reference_date = pd.Timestamp("2026-08-22")

features_df["Founded Year (Standardized)"] = pd.to_numeric(
    features_df["Founded Year (Standardized)"],
    errors="coerce"
)

features_df["Company Age (Years)"] = (
    reference_date.year - features_df["Founded Year (Standardized)"]
)

# Flag missing or implausible founding years for review
features_df["Company Age Validation"] = np.where(
    features_df["Founded Year (Standardized)"].isna(),
    "Missing Founding Year",
    np.where(
        (features_df["Founded Year (Standardized)"] < 1800) |
        (features_df["Founded Year (Standardized)"] > reference_date.year),
        "Review Founding Year",
        "Valid"
    )
)

print("Reference date:", reference_date.date())
print("Feature dataset shape:", features_df.shape)
print("\nCompany age validation:")
print(features_df["Company Age Validation"].value_counts(dropna=False))
print("\nCompany age summary:")
print(features_df["Company Age (Years)"].describe())
```

```python
# Day 5 - Step 5: Document derived features

new_features = pd.DataFrame([
    [
        "Company Age (Years)",
        "Company age in years as of 2026-08-22",
        "Derived numeric",
        "Calculated as 2026 minus Founded Year (Standardized); reference date fixed at 2026-08-22"
    ],
    [
        "Company Age Validation",
        "Flags missing or implausible founding years",
        "Validation flag",
        "Checks missing years and years before 1800 or after 2026; all 180 current records passed"
    ]
], columns=feature_catalogue.columns)

feature_catalogue = pd.concat(
    [feature_catalogue, new_features],
    ignore_index=True
)

print("Updated feature catalogue shape:", feature_catalogue.shape)
print(feature_catalogue.tail(2).to_string(index=False))
```

```python
# Day 5 - Step 6: Save final feature deliverables

processed_dir = project_dir / "data" / "processed"
processed_dir.mkdir(parents=True, exist_ok=True)

feature_dataset_path = (
    processed_dir / "Day5_Feature_Dataset.csv"
)
feature_catalogue_path = (
    processed_dir / "Day5_Feature_Catalogue.csv"
)

# Save updated files
features_df.to_csv(feature_dataset_path, index=False)
feature_catalogue.to_csv(feature_catalogue_path, index=False)

# Verify saved files
saved_features = pd.read_csv(feature_dataset_path)
saved_catalogue = pd.read_csv(feature_catalogue_path)

print("Feature dataset:", feature_dataset_path)
print("Saved dataset shape:", saved_features.shape)

print("\nFeature catalogue:", feature_catalogue_path)
print("Saved catalogue shape:", saved_catalogue.shape)

print("\nMissing values in saved dataset:",
      int(saved_features.isna().sum().sum()))

print("\nFeature names:")
print(saved_features.columns.tolist())
```

```python
# Day 5 - Step 7: Baseline-model readiness audit

target_candidates = [
    "Conversion Outcome",
    "Converted",
    "Lead Status",
    "Opportunity Won",
    "Quote Accepted",
    "Revenue"
]

available_targets = [
    col for col in target_candidates
    if col in df.columns
]

audit_results = [
    {
        "Check": "Feature dataset rows",
        "Result": len(features_df),
        "Status": "PASS"
    },
    {
        "Check": "Feature dataset columns",
        "Result": len(features_df.columns),
        "Status": "PASS"
    },
    {
        "Check": "Missing values in feature dataset",
        "Result": int(features_df.isna().sum().sum()),
        "Status": (
            "PASS"
            if features_df.isna().sum().sum() == 0
            else "REVIEW"
        )
    },
    {
        "Check": "Observed outcome target columns",
        "Result": ", ".join(available_targets) if available_targets else "None found",
        "Status": "PASS" if available_targets else "BLOCKED"
    },
    {
        "Check": "Conversion prediction baseline",
        "Result": (
            "Target available"
            if available_targets
            else "Cannot evaluate without an observed conversion label"
        ),
        "Status": "READY" if available_targets else "BLOCKED"
    },
    {
        "Check": "Original Prospect Score and Classification",
        "Result": "Existing score-derived fields; not validated conversion outcomes",
        "Status": "EXCLUDE AS CONVERSION TARGETS"
    }
]

baseline_audit = pd.DataFrame(audit_results)

print(baseline_audit.to_string(index=False))

audit_path = (
    project_dir / "data" / "processed"
    / "Day5_Baseline_Readiness_Audit.csv"
)

baseline_audit.to_csv(audit_path, index=False)
print("\nAudit saved:", audit_path)
```
