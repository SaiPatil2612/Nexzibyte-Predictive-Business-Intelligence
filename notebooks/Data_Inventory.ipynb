```python

# Add operational and analytical metadata to the data dictionary

def field_role(field):
    if field in [
        "Company Size Score", "B2B Fit Score",
        "Technology Signal Score", "Business Need Score",
        "Recent Business Signal Score", "Decision-Maker Score"
    ]:
        return "Scoring component"
    if field == "Prospect Score":
        return "Scored output"
    if field == "Prospect Classification":
        return "Classification output"
    if field in [
        "Website Validation", "Employee Validation",
        "Duplicate Flag", "Data Quality Status"
    ]:
        return "Quality / validation"
    if field in ["Source(s) Used", "Date Verified"]:
        return "Lineage / verification"
    return "Company / prospect attribute"

def predictive_use(field):
    if field == "Prospect Score":
        return "Potential prioritization feature; conversion prediction unvalidated"
    if field == "Prospect Classification":
        return "Existing label; threshold and business meaning require confirmation"
    if field.endswith("Score"):
        return "Existing scoring input; rubric and predictive value require validation"
    if field in ["Source(s) Used", "Date Verified"]:
        return "Supports traceability and freshness assessment"
    if field in [
        "Website Validation", "Employee Validation",
        "Duplicate Flag", "Data Quality Status"
    ]:
        return "Supports quality controls; not a business outcome"
    return "Potential segmentation or feature; usefulness requires validation"

def verification_status(field):
    if field == "Employee Count (Numeric)":
        return "Four records flagged for employee validation review"
    if field == "Date Verified":
        return "All rows contain 22-08-2026; meaning/history to confirm"
    if field == "Prospect Score":
        return "Component sum matches for all 180 rows; predictive validity unverified"
    if field == "Data Quality Status":
        return "176 Valid; 4 Needs Review"
    return "Present in source; independent verification not established by this audit"

data_dictionary_v2 = data_dictionary.copy()

data_dictionary_v2["Field Role"] = data_dictionary_v2["Source Field"].map(field_role)
data_dictionary_v2["Predictive Usefulness"] = data_dictionary_v2["Source Field"].map(predictive_use)
data_dictionary_v2["Source"] = "NexziByte_Final_Scored.csv"
data_dictionary_v2["Verification Status"] = data_dictionary_v2["Source Field"].map(verification_status)

dictionary_v2_path = output_dir / "Day2_Data_Dictionary_v2.csv"
data_dictionary_v2.to_csv(dictionary_v2_path, index=False)

print("Fields documented:", len(data_dictionary_v2))
print("Dictionary columns:", len(data_dictionary_v2.columns))
print("Enhanced dictionary saved:", dictionary_v2_path)
print("Original source CSV modified: No")

display(data_dictionary_v2.head(5))
```

```python
# Day 3: Date and Score Consistency Checks

# 1. Validate Date Verified
date_col = "Date Verified"

parsed_dates = pd.to_datetime(
    df_raw[date_col],
    format="%d-%m-%Y",
    errors="coerce"
)

print("DATE VALIDATION")
print("Unparseable dates:", parsed_dates.isna().sum())
print("Date values found:")
display(parsed_dates.value_counts(dropna=False))

# 2. Verify Prospect Score arithmetic
score_columns = [
    "Company Size Score",
    "B2B Fit Score",
    "Technology Signal Score",
    "Business Need Score",
    "Recent Business Signal Score",
    "Decision-Maker Score"
]

calculated_score = df_raw[score_columns].sum(axis=1)

score_mismatch = calculated_score != df_raw["Prospect Score"]

print("\nSCORE CONSISTENCY")
print("Score mismatches:", int(score_mismatch.sum()))

if score_mismatch.any():
    display(
        df_raw.loc[
            score_mismatch,
            ["Company Name", *score_columns, "Prospect Score"]
        ]
    )
else:
    print("All prospect scores match the sum of their six components.")

# 3. Check source-file integrity
with open(source_path, "rb") as file:
    source_hash_after_audit = hashlib.sha256(file.read()).hexdigest()

print("\nSOURCE INTEGRITY")
print("Source unchanged:", source_hash_before == source_hash_after_audit)
```

```python
# Day 3: Documented ETL Cleaning Rules

cleaning_rules = pd.DataFrame([
    {
        "Rule ID": "ETL-01",
        "Check": "Company Name",
        "Action": "Trim leading/trailing whitespace",
        "Reason": "Standardize text safely"
    },
    {
        "Rule ID": "ETL-02",
        "Check": "Website / Domain",
        "Action": "Trim whitespace; flag invalid URLs",
        "Reason": "Improve website field consistency"
    },
    {
        "Rule ID": "ETL-03",
        "Check": "Date Verified",
        "Action": "Parse using day-month-year format",
        "Reason": "Ensure consistent date representation"
    },
    {
        "Rule ID": "ETL-04",
        "Check": "Repeated company names",
        "Action": "Flag; do not automatically delete",
        "Reason": "Prevent accidental loss of records"
    },
    {
        "Rule ID": "ETL-05",
        "Check": "Data Quality Status",
        "Action": "Preserve existing status and review flags",
        "Reason": "Retain original audit findings"
    },
    {
        "Rule ID": "ETL-06",
        "Check": "Prospect Score",
        "Action": "Validate against six component scores",
        "Reason": "Maintain arithmetic consistency"
    }
])

display(cleaning_rules)
```

```python
# Day 3: ETL Cleaning Pipeline

import re
from urllib.parse import urlparse

# Create a working copy; never modify df_raw
df_clean = df_raw.copy(deep=True)

# 1. Trim whitespace from text columns
text_columns = df_clean.select_dtypes(
    include=["object", "string"]
).columns

for col in text_columns:
    df_clean[col] = df_clean[col].map(
        lambda value: value.strip()
        if isinstance(value, str)
        else value
    )

# 2. Parse verification dates without changing the raw data
df_clean["Date Verified"] = pd.to_datetime(
    df_clean["Date Verified"],
    format="%d-%m-%Y",
    errors="coerce"
)

# 3. Validate website/domain format
def is_valid_website(value):
    if not isinstance(value, str) or not value.strip():
        return False

    value = value.strip()
    candidate = value if "://" in value else "https://" + value
    parsed = urlparse(candidate)

    return (
        parsed.scheme in {"http", "https"}
        and bool(parsed.hostname)
        and "." in parsed.hostname
        and not any(char.isspace() for char in parsed.hostname)
    )

df_clean["ETL Website Format Valid"] = (
    df_clean["Website / Domain"].map(is_valid_website)
)

# 4. Check company-name uniqueness without deleting records
df_clean["ETL Company Name Unique"] = (
    ~df_clean["Company Name"].str.casefold().duplicated(keep=False)
)

# 5. Validate prospect score arithmetic
score_columns = [
    "Company Size Score",
    "B2B Fit Score",
    "Technology Signal Score",
    "Business Need Score",
    "Recent Business Signal Score",
    "Decision-Maker Score"
]

df_clean["ETL Score Consistent"] = (
    df_clean[score_columns].sum(axis=1)
    == df_clean["Prospect Score"]
)

# 6. Preserve existing Data Quality Status and Duplicate Flag
print("ETL TRANSFORMATION SUMMARY")
print("-" * 40)
print("Input rows:", len(df_raw))
print("Output rows:", len(df_clean))
print(
    "Invalid website formats:",
    int((~df_clean["ETL Website Format Valid"]).sum())
)
print(
    "Repeated company names:",
    int((~df_clean["ETL Company Name Unique"]).sum())
)
print(
    "Score inconsistencies:",
    int((~df_clean["ETL Score Consistent"]).sum())
)
print(
    "Existing review statuses preserved:",
    df_clean["Data Quality Status"].equals(
        df_raw["Data Quality Status"]
    )
)
print("Original DataFrame unchanged:", df_raw.equals(
    pd.read_csv(source_path)
))
```

```python
# Day 3: Save Clean Data and Data Quality Report

from pathlib import Path

clean_dir = project_dir / "data" / "clean"
processed_dir = project_dir / "data" / "processed"

# Ensure existing output directories are available
clean_dir.mkdir(parents=True, exist_ok=True)
processed_dir.mkdir(parents=True, exist_ok=True)

# Output file paths
clean_path = clean_dir / "Day3_Prospects_Clean.csv"
quality_path = processed_dir / "Day3_Data_Quality_Report.csv"
rules_path = processed_dir / "Day3_ETL_Cleaning_Rules.csv"

# Save clean layer
df_clean.to_csv(clean_path, index=False)

# Build data quality report
quality_checks = [
    ("Input row count", len(df_raw), "Informational"),
    ("Output row count", len(df_clean), "Must equal input"),
    ("Missing values in source", int(df_raw.isna().sum().sum()), "Review"),
    ("Exact duplicate source rows", int(df_raw.duplicated().sum()), "Review"),
    ("Invalid website formats", int((~df_clean["ETL Website Format Valid"]).sum()), "Review"),
    ("Repeated company names", int((~df_clean["ETL Company Name Unique"]).sum()), "Review"),
    ("Score inconsistencies", int((~df_clean["ETL Score Consistent"]).sum()), "Must be zero"),
    ("Date parsing failures", int(df_clean["Date Verified"].isna().sum()), "Review"),
    ("Existing Needs Review records", int((df_clean["Data Quality Status"] == "Needs Review").sum()), "Preserve and investigate"),
]

quality_report = pd.DataFrame(
    quality_checks,
    columns=["Check", "Result Count", "Acceptance Rule"]
)

quality_report.to_csv(quality_path, index=False)
cleaning_rules.to_csv(rules_path, index=False)

# Verify saved outputs
saved_clean = pd.read_csv(clean_path)

print("DAY 3 OUTPUTS SAVED")
print("-" * 45)
print("Clean dataset:", clean_path)
print("Quality report:", quality_path)
print("Cleaning rules:", rules_path)
print("Saved dataset rows:", len(saved_clean))
print("Saved dataset columns:", len(saved_clean.columns))
print("All three files exist:", all(
    path.exists() for path in [clean_path, quality_path, rules_path]
))

display(quality_report)
```
